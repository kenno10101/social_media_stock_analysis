## 3) Everything necessary to know (tools, usernames & passwords, etc.)

### Required software / tools

* **Python 3.11+**
* **Docker Desktop** (to run Kafka and MongoDB)
* **Java 11+** (required for Spark / PySpark)

### Docker services used (from `docker-compose.yml`)

* Zookeeper: `confluentinc/cp-zookeeper:7.6.0` (port `2181`)
* Kafka: `confluentinc/cp-kafka:7.6.0` (port `9092`)
* MongoDB: `mongo:latest` (port `27017`)
* Mongo Express: `mongo-express:latest` (port `8081`)
